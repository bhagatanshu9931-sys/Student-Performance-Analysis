```python

import pandas as pd

df = pd.read_csv("student-mat.csv", sep=";")

print("Dataset Shape:", df.shape)

print("\nColumn Names:")
print(df.columns)

print("\nData Types:")
print(df.dtypes)

print("\nMissing Values:")
print(df.isnull().sum())
```

```python

print("Duplicates:", df.duplicated().sum())


df = df.drop_duplicates()


print("Missing Values:")
print(df.isnull().sum())

print("\nCleaned Dataset Shape:", df.shape)
```

```python
# Student Performance Analysis

print("Average Final Grade:", round(df["G3"].mean(), 2))

print("Highest Final Grade:", df["G3"].max())

print("Lowest Final Grade:", df["G3"].min())

print("Students Scoring Above 15:", (df["G3"] > 15).sum())

print("Students Scoring Below 10:", (df["G3"] < 10).sum())

print("\nAverage Grade by Gender:")
print(df.groupby("sex")["G3"].mean())
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(df["G3"], bins=10, kde=True)

plt.title("Distribution of Student Final Grades")
plt.xlabel("Final Grade")
plt.ylabel("Number of Students")

plt.show()
```

```python
plt.figure(figsize=(7, 5))

sns.barplot(data=df, x="sex", y="G3")

plt.title("Average Final Grade by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Final Grade")

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("student-mat.csv", sep=";")

print("Setup completed successfully!")
```

```python
plt.figure(figsize=(8, 5))

sns.barplot(data=df, x="studytime", y="G3")

plt.title("Study Time vs Final Grades")
plt.xlabel("Weekly Study Time")
plt.ylabel("Average Final Grade")

plt.show()
```

```python
print("Average Grade Based on Study Time:")

print(df.groupby("studytime")["G3"].mean().round(2))
```

```python
plt.figure(figsize=(10, 7))

correlation = df[["age", "studytime", "failures",
                  "absences", "G1", "G2", "G3"]].corr()

sns.heatmap(correlation, annot=True, cmap="coolwarm", fmt=".2f")

plt.title("Correlation Between Student Performance Factors")

plt.show()
```

```python
print("Correlation between Study Time and Final Grade:")
print(df["studytime"].corr(df["G3"]).round(3))
```

```python
plt.figure(figsize=(8, 5))

grade_data = df[["G1", "G2", "G3"]].mean()

sns.barplot(x=grade_data.index, y=grade_data.values)

plt.title("Comparison of First, Second and Final Grades")
plt.xlabel("Exam")
plt.ylabel("Average Grade")

plt.show()
```
